```python
import os
import boto3
from dotenv import load_dotenv

# 1. On spécifie que le fichier .env se trouve dans le dossier parent (..)
load_dotenv(dotenv_path="../.env")

# 2. Vérification rapide pour s'assurer que les variables sont bien chargées
print("AWS_ACCESS_KEY_ID :", os.getenv('AWS_ACCESS_KEY_ID'))
print("S3_ENDPOINT_URL :", os.getenv('S3_ENDPOINT_URL'))

# 3. Initialiser le client S3
s3_client = boto3.client(
    's3',
    aws_access_key_id=os.getenv('AWS_ACCESS_KEY_ID'),
    aws_secret_access_key=os.getenv('AWS_SECRET_ACCESS_KEY'),
    region_name=os.getenv('AWS_REGION'),
    endpoint_url=os.getenv('S3_ENDPOINT_URL')
)

bucket_name = os.getenv('S3_BUCKET_NAME')

# Chemin local du fichier brut
chemin_local = "../data/bronze/stocks/attijari_3ans.xlsx" 

# Chemin de destination dans le bucket MinIO
chemin_s3 = "bronze/stocks/attijari_3ans.xlsx"

# Upload du fichier
try:
    s3_client.upload_file(chemin_local, bucket_name, chemin_s3)
    print(f"✅ Fichier uploadé avec succès sur MinIO : s3://{bucket_name}/{chemin_s3}")
except Exception as e:
    print(f"❌ Erreur lors de l'upload : {e}")
```
